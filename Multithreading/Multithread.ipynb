```python
from threading import Thread
def task():
    print("Thread is running")
    
# creating Thread
t = Thread(target=task)

#start thread
t.start()

# waiting for completion
t.join()
print("Main Program Ends")
```

```python
from threading import Thread
from time import sleep
def task():
    for i in range(3):
        print(i)
        sleep(1)
    
# creating Thread
t = Thread(target=task)

#start thread
t.start()

# waiting for completion
t.join()
print("Main Program Ends")
```
